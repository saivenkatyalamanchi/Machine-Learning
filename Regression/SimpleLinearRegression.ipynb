```python
# simple linear regression is just a model with one feature. i.e y = w*x + b 
```

```python
import numpy as np
```

```python
x_values = np.random.randint(-100,100,50)
x_values
```

```python
y_values = 3*x_values - 5
y_values
```

```python
# so w = 3 and b = -5
```

```python
w, b = 0, 0
```

```python
epochs = 100000
alpha = 0.0001
n = len(x_values)
for epoch in range(epochs):
    loss = 0
    dw = 0
    db = 0
    for i in range(n):
        y_pred = w*x_values[i] + b
        loss += (y_pred - y_values[i]) ** 2
        dw += (2/n) * (y_pred - y_values[i]) * x_values[i]
        db += (2/n) * (y_pred - y_values[i])
    loss = loss/n
    if epoch % 100 == 0:
        print(f"Loss = {loss}")
    w = w - alpha * dw
    b = b - alpha * db

# implemented batch gradient descent also try schochastic and mini batch gradient descent also. 
# schochastic -> we updat w,b for each training example
# mini batch -> we divide training set into batches and update w,b only when a batch is processed.
```

```python
print(f"final values of learned parameters w and b are {w} and {b} respectively.")
```
